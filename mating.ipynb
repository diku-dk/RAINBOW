```python
import matplotlib.pyplot as plt
import numpy as np
import polyscope as ps

from stl import mesh

import rainbow.math.involute as INVOLUTE
import rainbow.math.quaternion as Q

from rainbow.procedural.gears.types import Gear, GearSpec, PlanetaryGearSpec
from rainbow.procedural.gears.generators import GearFactory

ps.init()

plt.rc('font', size=12)
plt.rc('text', usetex=True)
plt.rc('font', family='serif')

def register_gear(name: str, gear: Gear) -> ps.SurfaceMesh:
    V = Q.rotate_array(gear.orientation, gear.mesh.V) + gear.position
    T = gear.mesh.T
    return ps.register_surface_mesh(name, V, T)
```

```python
z_sun = 33
z_planet_options = [i for i, x in enumerate(((z_sun + z_internal) / 3).is_integer() for z_internal in (z_sun + 2 * i for i in range(100))) if x and i > 4]
print(f'{z_planet_options=}')
```

```python
factory = GearFactory()

m = 1
#z_sun = 5
z_planet = 18
z_internal = z_sun + 2 * z_planet
print(f'{z_internal=}')

planetary_spec = PlanetaryGearSpec(m, z_sun, z_planet, z_internal)
planetary_gear = factory.create_planetary_gear(planetary_spec, 10)

register_gear("sun_gear", planetary_gear.sun_gear)
for i, gear in enumerate(planetary_gear.planet_gears):
    register_gear(f"planet_{i}", gear)
register_gear("ring_gear", planetary_gear.ring_gear)

ps.show()
```
