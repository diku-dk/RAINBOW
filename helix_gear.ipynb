```python
import rainbow.math.involute as INV
import rainbow.math.vector3 as V3
import rainbow.math.quaternion as Q
import rainbow.simulators.prox_rigid_bodies.api as API
import rainbow.geometry.surface_mesh as MESH
from rainbow.simulators.prox_rigid_bodies.types import Engine

import rainbow.procedural.gears.generators as GEAR
import rainbow.procedural.gears.mating as MATING

import rainbow.procedural.gears as GEAR

import polyscope as ps
import numpy as np

helix_angle = 0

ps.init()

factory = GEAR.GearFactory(100, 1, 1)
spec1 = GEAR.GearSpec(1, 19, helix_angle=helix_angle, bevel_cone_angle=None, is_internal=False)
spec2 = GEAR.GearSpec(1, 20, helix_angle=helix_angle, bevel_cone_angle=None, is_internal=False)
gear1 = factory.create_gear(spec1, 5, subdivisions=0)
gear2 = factory.create_gear(spec2, 5, subdivisions=0)

gear_mesh1 = gear1.mesh
gear_mesh2 = gear2.mesh

theta = np.deg2rad(55)
roll_angle = (spec1.rp / spec2.rp + 1) * theta
print(np.rad2deg(roll_angle))
phi = roll_angle + gear2.spec.gamma_p
R = np.array([
    [np.cos(phi), -np.sin(phi), 0],
    [np.sin(phi), np.cos(phi), 0],
    [0, 0, 1],
]).T

gear_mesh2.V = gear_mesh2.V @ R

#offset = spec1.ra + spec2.ra + 1
offset = spec1.rp + spec2.rp
gear_mesh2.V += offset * np.array([np.cos(theta), np.sin(theta), 0])

midpoint = offset / 2

ps.register_surface_mesh("gear1", gear_mesh1.V, gear_mesh1.T)
ps.register_surface_mesh("gear2", gear_mesh2.V, gear_mesh2.T)
ps.set_ground_plane_mode("none")
ps.set_view_projection_mode("orthographic")
ps.look_at([midpoint, 0, 1], [midpoint, 0, 0])
ps.show()
```
